```python
""" gaze and events feather to csv"""

import os
import pandas as pd

def convert_feathers_to_csv(input_dir, output_dir):
    os.makedirs(output_dir, exist_ok=True)

    for filename in os.listdir(input_dir):
        if filename.endswith('.feather'):
            input_path = os.path.join(input_dir, filename)
            output_filename = filename.replace('.feather', '.csv')
            output_path = os.path.join(output_dir, output_filename)

            try:
                print(f"Converting: {filename} -> {output_filename}")
                df = pd.read_feather(input_path)
                df.to_csv(output_path, index=False)
            except Exception as e:
                print(f"❌ Failed to convert {filename}: {e}")

    print(f"✅ Done: {input_dir} → {output_dir}")

if __name__ == '__main__':
    folder_pairs = [
        (r"Z:\BioMotionAnlyze\analyze\data\pymovement data\202511_typeA\rt\events", r"Z:\BioMotionAnlyze\analyze\data\pymovement data\202511_typeA\rt\events_csv"),
        (r"Z:\BioMotionAnlyze\analyze\data\pymovement data\202511_typeA\rt\preprocessed", r"Z:\BioMotionAnlyze\analyze\data\pymovement data\202511_typeA\rt\gaze_csv"),
    ]

    for input_dir, output_dir in folder_pairs:
        print(f"\n📁 Processing folder: {input_dir}")
        convert_feathers_to_csv(input_dir, output_dir)

    print("\n🌟 All conversions complete.")
```

```python
"""gaze .csv make pixel, position, velocity columns from [x y] to x, y"""

import os
import pandas as pd
import ast
import re

def safe_parse_position(val):
    if isinstance(val, list) and len(val) == 2:
        return val
    if isinstance(val, str):
        try:
            fixed = re.sub(r'(?<=\d)\s+(?=-?\d)', ', ', val.strip())
            parsed = ast.literal_eval(fixed)
            if isinstance(parsed, list) and len(parsed) == 2:
                return parsed
        except:
            return [None, None]
    return [None, None]

def split_and_replace(df, col_name):
    if col_name in df.columns:
        parsed = df[col_name].apply(safe_parse_position)
        df[f"{col_name}_x"] = parsed.apply(lambda p: p[0])
        df[f"{col_name}_y"] = parsed.apply(lambda p: p[1])
        df.drop(columns=[col_name], inplace=True)
    return df

def process_gaze_folder(input_dir, output_dir):
    os.makedirs(output_dir, exist_ok=True)
    files = [f for f in os.listdir(input_dir) if f.endswith(".csv")]

    for filename in files:
        input_path = os.path.join(input_dir, filename)
        output_path = os.path.join(output_dir, filename)

        try:
            df = pd.read_csv(input_path)

            for col in ["pixel", "position", "velocity"]:
                df = split_and_replace(df, col)

            df.to_csv(output_path, index=False)
            print(f"处理完成: {filename}")

        except Exception as e:
            print(f"处理失败: {filename}，错误信息：{e}")

    print(f"\n🌟 所有 gaze 文件处理完成，输出目录：{output_dir}")

# ========= 主执行入口 =========
if __name__ == '__main__':
    input_dir = r"Z:\BioMotionAnlyze\analyze\data\pymovement data\202511_typeA\rt\gaze_csv"
    output_dir = r"Z:\BioMotionAnlyze\analyze\data\pymovement data\202511_typeA\rt\gaze_processed"

    process_gaze_folder(input_dir, output_dir)
```

```python
"""events .csv make pixel, position, velocity columns from [x y] to x, y"""

import os
import pandas as pd
import ast
import re

def safe_parse_position(val):
    if isinstance(val, list) and len(val) == 2:
        return val
    if isinstance(val, str):
        try:
            fixed = re.sub(r'(?<=\d)\s+(?=-?\d)', ', ', val.strip())
            parsed = ast.literal_eval(fixed)
            if isinstance(parsed, list) and len(parsed) == 2:
                return parsed
        except:
            return [None, None]
    return [None, None]

def split_and_replace(df, col_name):
    if col_name in df.columns:
        parsed = df[col_name].apply(safe_parse_position)
        df[f"{col_name}_x"] = parsed.apply(lambda p: p[0])
        df[f"{col_name}_y"] = parsed.apply(lambda p: p[1])
        df.drop(columns=[col_name], inplace=True)
    return df

def process_events_folder(input_dir, output_dir):
    os.makedirs(output_dir, exist_ok=True)
    files = [f for f in os.listdir(input_dir) if f.endswith(".csv")]

    for filename in files:
        input_path = os.path.join(input_dir, filename)
        output_path = os.path.join(output_dir, filename)

        try:
            df = pd.read_csv(input_path)
            df = split_and_replace(df, "location")
            df.to_csv(output_path, index=False)
            print(f"处理完成: {filename}")
        except Exception as e:
            print(f"处理失败: {filename}，错误信息：{e}")

    print(f"\n🌟 所有 events 文件处理完成，输出目录：{output_dir}")

# ========= 主执行入口 =========
if __name__ == '__main__':
    input_dir = r"Z:\BioMotionAnlyze\analyze\data\pymovement data\202511_typeA\rt\events_csv"
    output_dir = r"Z:\BioMotionAnlyze\analyze\data\pymovement data\202511_typeA\rt\events_processed"
    
    process_events_folder(input_dir, output_dir)
```

```python
"""delete fixation of fix at start trial fixation point"""
import os
import pandas as pd

# === 路径设置 ===
input_dir = r"Z:\BioMotionAnlyze\analyze\data\pymovement data\exp 202504\events_processed"

# === Fixation start box 设置 ===
start_box_centers = [(0, 0)] # legacy config
# start_box_centers = [(-10, 10), (-10, -10), (10, 10), (10, -10)] # new config
half_box_size = 2.5  # 边长为5的一半

def is_in_start_box(x, y):
    for cx, cy in start_box_centers:
        if (cx - half_box_size <= x <= cx + half_box_size) and (cy - half_box_size <= y <= cy + half_box_size):
            return True
    return False

# === 批量处理 ===
for fname in os.listdir(input_dir):
    if not fname.endswith(".csv"):
        continue

    file_path = os.path.join(input_dir, fname)
    df = pd.read_csv(file_path)

    # 找到第一个 fixation 行
    fixation_idx = df[df["name"] == "fixation"].index
    if len(fixation_idx) == 0:
        continue  # 没有fixation则跳过

    first_fix_idx = fixation_idx[0]
    first_fix = df.loc[first_fix_idx]

    # 判断是否落在起始点框内
    if is_in_start_box(first_fix["location_x"], first_fix["location_y"]):
        df = df.drop(first_fix_idx)  # 删除该 fixation 行
        df = df.reset_index(drop=True)  # 重新索引
        print(f"Removed first fixation in: {fname}")

    # 保存回原路径（覆盖）
    df.to_csv(file_path, index=False)
```
